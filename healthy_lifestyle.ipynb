# Lifestyle and Health #
This project examines the relationship of different attributes like age, gender, bmi etc of a person along with the lifestyle choices like daily steps, sleep hours, water intake, calories consumed, smoking status, alcohol usage and genetic factors like family history, disease risk with the health indicators such as resting heart rate, systolic and diastolic blood pressure and cholesterol.
The dataset is obtained from the following link:
https://www.kaggle.com/datasets/chik0di/health-and-lifestyle-dataset

```python
#importing necessary libraries
import pandas as pd
#read data
health_data = pd.read_csv('health_lifestyle_dataset.csv')
#print the column headings
health_data.columns
```

```python
#information about the data
health_data.info()
```

```python
#summary statistics
health_data.describe(include='all')
```

```python
health_data = health_data.dropna()
health_data.shape
```

 Since this is a very large data, let us take a sample of 400 which is .4 percent of the whole population

```python
sample_data = health_data.sample(n= int(health_data.shape[0]*0.004))
sample_data.info()
```

```python
#import necessary libraries for visualization
import seaborn as sns
import matplotlib.pyplot as plt
sns.boxplot(x='disease_risk', y = "resting_hr", hue = 'smoker' , data = sample_data)
plt.show()
```

```python
data1 = health_data[['age','bmi', 'daily_steps', 'sleep_hours',
       'water_intake_l', 'calories_consumed', 'smoker', 'alcohol',
       'resting_hr', 'systolic_bp', 'diastolic_bp', 'cholesterol',
       'family_history', 'disease_risk']]
plt.figure(figsize = (8,6))
data1.corr()
sns.heatmap(data1.corr(), cmap = 'viridis')
```

Analysis: This shows there is no correlation between the numerical variables in this datasets which makes us doubt whether this
is a data collected from real life. 

Let us analyze whether the categorical variable which is gender has any correlation with the other variables. To check the relationship 9 numerical variables are included in the list named 'factors'


```python

factors = ['bmi', 'daily_steps', 'sleep_hours', 'water_intake_l', 'calories_consumed', 'resting_hr', 'systolic_bp', 'diastolic_bp', 'cholesterol'
      ]
fig = plt.figure(figsize = (10, 6))
plot = 0
for factor in factors:
    plot += 1
    ax = plt.subplot(3, 3, plot)
    plt.subplots_adjust(wspace = .95)
    #sns.scatterplot(x= factor, y= 'resting_hr',  data = sample_data)
    sns.boxplot(x=factor, hue= 'gender', data = sample_data)
    plt.legend(bbox_to_anchor=(1,1))
plt.tight_layout()   
    
plt.show()
```

Analysis: There is no specific pattern observable from these plots. Most variables are less for females compared to males and some variables
like water intake is higher for females. 

Let us compare gender with the following variables in the list 'factors'

```python
factors = [ 'smoker', 'alcohol','family_history', 'disease_risk']
fig = plt.figure(figsize = (10, 6))
plot = 0
for factor in factors:
    plot += 1
    ax = plt.subplot(3, 3, plot)
    plt.subplots_adjust(wspace = .95)
    sns.countplot(x=factor, hue= 'gender', data = sample_data)
    plt.legend(bbox_to_anchor=(1,1))
plt.tight_layout()   
    
plt.show()
```

```python
categories = ['male', 'female']
#to find how many of the smokers in the sample are male and female
sample_data.gender[sample_data.smoker==1].value_counts()
```

To find what percentage is male vs. female among smokers, alcohol users, family history and disease risk

```python

factors = [ 'smoker', 'alcohol','family_history', 'disease_risk']
fig = plt.figure(figsize = (10, 6))
plot = 0
for factor in factors:
    print(factor)
    plot += 1
    ax = plt.subplot(3, 3, plot)
    plt.subplots_adjust(wspace = .95)
    #sns.scatterplot(x= factor, y= 'resting_hr',  data = sample_data)
    print(health_data.gender[health_data[factor] == 0].value_counts())
    factor_data = health_data.gender[health_data[factor] == 0].value_counts(normalize = True)
    plt.pie(factor_data, labels =['male', 'female'], autopct ='%0.1f%%')
    plt.title(factor)
    
    #plt.legend(bbox_to_anchor=(1,1))
plt.tight_layout()   
    
plt.show()
```

```python
Analysis: Almost 50 percent of the smokers and alcohol users are females. The same division for disease risk and family history. This again shows that 
this data may not be taken from real life.
```
